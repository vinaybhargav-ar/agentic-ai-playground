```python
from langchain.tools import tool
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent

from google.colab import userdata
```

## Create LLM

```python
llm = None
try:
    llm = ChatOpenAI(
        model="gpt-4o",
        api_key=userdata.get("OPENAI_APIKEY"),
        temperature=0
    )
except Exception as e:
    print(f"Error creating model: {e}")
```

## Create a tool which the agent can use.

Note the tool description is required and the `@tool` decorator is what makes this a tool.

```python
@tool
def calculate(num1: int, num2: int, operation: str) -> float:
  """
  Performs addition, subtraction, multiplication or division.
  """
  print(f"Using the tool!")
  if operation == 'add':
    return num1 + num2

  if operation == 'subtract':
    return num1 - num2

  if operation == 'multiply':
    return num1 * num2

  if operation == 'divide':
    return num1 / num2

  return 0
```

## Create Agent

```python
agent = create_agent(
    model=llm,
    tools=[calculate],
    system_prompt="You are a helpful calculator assistant. Use the calculate tool for arithmetic."
)
```

## Invoking Agent

```python
def run_agent(prompt: str):
  """
  The main execution loop for the Calculator Agent.
  """
  print(f"\n--- Running the Calculator Agent ---")

  try:
    result = agent.invoke(
        {
            "messages": [
                {
                    "role": "user",
                    "content": prompt
                }
            ]
        }
    )

    return result["messages"][-1].content

  except Exception as e:
    return f"Agent error: {e}"
```

## Calling the Agent to test

```python
answer = run_agent("what is 100 multiplied by 20?")
print(answer)
```

```python
run_agent("What is the capital of France?")
```
